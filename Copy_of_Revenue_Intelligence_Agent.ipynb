```python
with open("project/data/deals.json") as f:
    data = json.load(f)

deals = data["deals"]
```

```python
expected_revenue = 0

for deal in deals:
    expected_revenue += deal["value"] * deal["probability"]

print(expected_revenue)
```

```python
inactive_deals = []

for deal in deals:
    if deal["last_activity_days"] > 7:
        inactive_deals.append(deal)
```

```python
issues = [
    f"{len(inactive_deals)} deals inactive for more than 7 days"
]
```

```python
from jinja2 import Template
```

```python
with open("project/templates/report_template.html") as f:
    template = Template(f.read())
```

```python
final_html = template.render(
    target_revenue=data["target_revenue"],
    expected_revenue=round(expected_revenue),
    issues=issues
)
```

```python
with open("project/output/report.html", "w") as f:
    f.write(final_html)
```
